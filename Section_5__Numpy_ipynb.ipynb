```python
# Indexing
# NumPy uses zero-based indexing, just like Python lists :
x = np.array([10, 20, 30, 40])

print(x[0])
print(x[2])
```

```python
# 2D Array Indexing :
x = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(x[1 ,1])
```

```python
# Array Operations :
x = np.array([1, 2, 3, 4])

print(x + 10)
print(x * 2)
print(x ** 2)
```

```python
# Operations Between Arrays :
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print(a + b)
print(a - b)
print(a * b)
```

```python
# Statistics
# NumPy is also useful for basic statistics :
grades = np.array([70, 80, 90, 100])

print(np.mean(grades))
print(np.min(grades))
print(np.max(grades))
```

```python
#  Matrix Multiplication :
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])


print(A * B)

print('*' * 90 )

print(A @ B )
```

```python
# Dot Product :
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

dot_result = np.dot(a , b)

print(dot_result)
```
