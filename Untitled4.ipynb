```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def evaluate_structural_health():
    # Simulation parameters representing sensor nodes across bridge sections
    np.random.seed(42)
    sensors = ['Pier A-1', 'Pier A-2', 'Main Span B', 'Pier C-1', 'Pier C-2']

    # Normal baseline ultrasonic pulse velocity (UPV) for healthy concrete (m/s)
    baseline_speed = 4200.0

    # Simulate current readings with a critical micro-crack anomaly introduced in Main Span B
    current_speeds = [
        baseline_speed + np.random.normal(0, 30),
        baseline_speed + np.random.normal(0, 25),
        baseline_speed * 0.78,  # Significant drop indicating internal cracking/risk
        baseline_speed + np.random.normal(0, 35),
        baseline_speed + np.random.normal(0, 20)
    ]

    # Calculate percentage deviation and assess structural integrity status
    deviations = [(baseline_speed - speed) / baseline_speed * 100 for speed in current_speeds]
    statuses = []
    actions = []

    for dev in deviations:
        if dev > 15.0:
            statuses.append("CRITICAL RISK (Internal Cracking)")
            actions.append("Immediate Inspection Required")
        elif dev > 8.0:
            statuses.append("MODERATE WARNING")
            actions.append("Schedule Maintenance")
        else:
            statuses.append("NORMAL / SECURE")
            actions.append("No Action Needed")

    # Compile comprehensive engineering report dataframe
    report_df = pd.DataFrame({
        'Sensor Node': sensors,
        'Baseline (m/s)': [baseline_speed] * 5,
        'Measured UPV (m/s)': [round(s, 2) for s in current_speeds],
        'Velocity Drop (%)': [round(d, 2) for d in deviations],
        'Structural Status': statuses,
        'Recommended Action': actions
    })

    print("=" * 95)
    print(" EXPERT NON-DESTRUCTIVE TESTING (NDT) - STRUCTURAL HEALTH MONITORING REPORT")
    print(" Application: Ultrasonic Pulse Velocity (UPV) & Acoustic Velocity Analysis for Bridges")
    print("=" * 95)
    print(report_df.to_string(index=False))
    print("=" * 95)

    # Generate an executive visualization chart suitable for administrative review
    plt.figure(figsize=(10, 5))
    colors = ['#2ca02c' if dev <= 8 else ('#ff7f0e' if dev <= 15 else '#d62728') for dev in deviations]

    bars = plt.bar(sensors, current_speeds, color=colors, alpha=0.85, edgecolor='black', width=0.55)
    plt.axhline(y=baseline_speed, color='blue', linestyle='--', linewidth=2, label=f'Safe Baseline ({baseline_speed} m/s)')
    plt.axhline(y=baseline_speed * 0.85, color='red', linestyle=':', linewidth=2, label='Critical Safety Threshold (15% Drop)')

    plt.title('Bridge & Infrastructure Structural Health Monitoring (UPV Analysis)', fontsize=14, fontweight='bold', pad=15)
    plt.xlabel('Sensor Monitoring Stations', fontsize=12, fontweight='bold')
    plt.ylabel('Ultrasonic Pulse Velocity (m/s)', fontsize=12, fontweight='bold')
    plt.ylim(3000, 4600)
    plt.grid(axis='y', linestyle='--', alpha=0.5)
    plt.legend(loc='lower left', fontsize=10)

    # Add precise values on top of bars
    for bar, speed in zip(bars, current_speeds):
        plt.text(bar.get_x() + bar.get_width()/2.0, speed + 30, f"{speed:.1f} m/s", ha='center', va='bottom', fontsize=10, fontweight='bold')

    plt.tight_layout()
    plt.show()

evaluate_structural_health()
```
